```python
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine learning preprocessing
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

# Train-test split
from sklearn.model_selection import train_test_split

# Machine Learning Algorithms
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv(r"C:\Users\NG_\Agritech-connect-portfolio\merged_crop_dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.rename(columns={'label':'crop'}, inplace=True)
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.describe()
```

```python
#Visualize crop distribution
plt.figure(figsize=(15,6))

sns.countplot(
    x=df.iloc[:, -1],
    data=df
)

plt.xticks(rotation=90)
plt.title("Crop Distribution")
plt.show()
```

```python
#Identifying feature columns and target
x = df.drop(df.columns[-1], axis=1)
y = df[df.columns[-1]]
```

```python
#Encode crop labels
encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

```python
#Scalling features
scaler =StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
#Split dataset
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled, y, test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)

print(y_train.shape)
print(y_test.shape)
```

```python
#Creating models dictionary
models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "Support Vector Machine": SVC(),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Gaussian Naive Bayes": GaussianNB()
}
```

```python
results = []
```

```python
#Train models
for name, model in models.items():
    print(f"Model: {name}")
```

```python
model.fit(x_train, y_train)
```

```python
#Make predictions
y_pred = model.predict(x_test)
```

```python
#Calculate metrics
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,y_pred,average='weighted')

recall = recall_score(
    y_test,y_pred,average='weighted')

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
```

```python
#save results
results.append([name,accuracy,precision,recall])
```

```python
#create results table
results_df = pd.DataFrame(
    results,
    columns=["Model", "Accuracy", "Precision", "Recall"])

results_df
```

```python
#sort best models
results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df
```

```python
#Visualize accuracy
plt.figure(figsize=(12,6))

sns.barplot(
    x="Model",
    y="Accuracy",
    data=results_df
)

plt.xticks(rotation=45)
plt.title("Model Accuracy Camparison")
plt.show()
```

```python
#Select best model
best_model = RandomForestClassifier()
```

```python
#Train best model again
best_model.fit(x_train, y_train)
```

```python
#Make predictions
y_pred = best_model.predict(x_test)
```

```python
#Create confusion matrix
cm = confusion_matrix(
    y_test,y_pred
)
print(cm)
```

```python
#Visualize confusion matrix
plt.figure(figsize=(12,8))

sns.heatmap(
    cm,
    annot=True,
    fmt='d'
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#Classification report
print(classification_report(y_test,y_pred))
```

```python
#Check accuracy
accuracy = accuracy_score(
    y_test,y_pred)

print("Final Accuracy:", accuracy)
```

```python
#Compare with results table
results_df
```

Random Forest achieved the highest accuracy among all machine learning models tested. The confusion matrix showed strong classification performance with minimal misclassification. Classification report metrics indicated high precision, recall, and F1-score, making Random Forest the preferred model for crop prediction.
